```python
import pandas as pd
missing_counts = df.isna().sum()
print("Missing values per column:", missing_counts)
print("Total missing cells:", df.isna().sum().sum())
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

```python
df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
df = df.drop(columns=['customerID'])
df = df.drop_duplicates()

print(df.isna().sum())
print(df.isna().sum().sum())
print(df.duplicated().sum())
print(df.shape)
```

```python
df.describe()
```

```python
q1 = df['MonthlyCharges'].quantile(0.25)
q3 = df['MonthlyCharges'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers = df[
    (df['MonthlyCharges'] < lower_bound) | (df['MonthlyCharges'] > upper_bound)
]
print(len(outliers))
```

```python
numeric_cols = df.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = df.select_dtypes(exclude=["number"]).columns.tolist()

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

```python
print("Final shape of cleaned dataset:", df.shape)
print("\nMissing values remaining:")
print(df.isna().sum())
print("\nDuplicate rows remaining:", df.duplicated().sum())
df.head(10)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('/content/sample_data/project.csv')

categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
  df[col] = df[col].astype(str).str.strip()

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

df['Churn_Numeric'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

df = df.drop(columns=['customerID'])
df = df.drop_duplicates()

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)

df.head()
```

```python
tenure_counts = df['tenure'].value_counts().sort_index()

plt.figure(figsize=(8, 4))
plt.plot(tenure_counts.index, tenure_counts.values, color='#aaafff', marker='*')
plt.title('Customers Over Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Number of Customers')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['MonthlyCharges'], kde=True, color='skyblue')
plt.title('Distribution of Monthly Charges')
plt.xlabel('Monthly Charges')
plt.ylabel('Frequency')
plt.show()
```

```python
daily_charges = df.groupby('tenure')['MonthlyCharges'].mean()

plt.figure(figsize=(8, 4))
plt.plot(daily_charges.index, daily_charges.values, color='#afafaf', marker='o')
plt.title('Average Monthly Charges Over Tenure')
plt.xlabel('Tenure (Months)')
plt.ylabel('Average Monthly Charges')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
contract_counts = df['Contract'].value_counts()

plt.figure(figsize=(8,4))
plt.bar(contract_counts.index, contract_counts.values, color='#f666aa')
plt.title('Number of Customers per Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
avg_charges_contract = df.groupby('Contract')['MonthlyCharges'].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(avg_charges_contract.index, avg_charges_contract.values, color='#b4466a')
plt.title('Average Monthly Charges per Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Average Monthly Charges')
plt.show()
```

```python
avg_charges_method = df.groupby('PaymentMethod')['MonthlyCharges'].mean().sort_values()

plt.figure(figsize=(10, 4))
plt.barh(avg_charges_method.index, avg_charges_method.values, color='#4B2110')
plt.title('Average Monthly Charges per Payment Method')
plt.xlabel('Average Monthly Charges')
plt.ylabel('Payment Method')
plt.show()
```

```python
payment_counts = df['PaymentMethod'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(payment_counts.values,
        labels=payment_counts.index,
        autopct='%1.0f%%',
        colors=['#CC3A63', '#F9F0E0', '#BC4F4F', '#D98282'],
        startangle=140)

plt.title('Payment Method Share')
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.histplot(df['MonthlyCharges'], bins=15, color='#810B10')
plt.title('Distribution of Monthly Charges')
plt.xlabel('Monthly Charges')
plt.ylabel('Number of Customers')
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.violinplot(x='PaymentMethod', y='MonthlyCharges', data=df, color='#760031')
plt.title('Monthly Charges Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Monthly Charges')
plt.show()
```

```python
plt.figure(figsize=(12,5))
for contract in df['Contract'].unique():
    subset = df[df['Contract'] == contract]
    sns.kdeplot(subset['MonthlyCharges'], label=contract)
plt.title('Monthly Charges Density by Contract Type')
plt.xlabel('Monthly Charges')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,6))
numeric_cols = df[['tenure', 'MonthlyCharges', 'TotalCharges']]
sns.heatmap(numeric_cols.corr(), annot=True, cmap='Reds')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sample_df = df.sample(400, random_state=1)
sns.pairplot(sample_df[['tenure', 'MonthlyCharges', 'TotalCharges']],
             diag_kind='hist', plot_kws={'alpha':0.4, 'color': '#5D3140'})
plt.suptitle('Pair Plot: Tenure, Monthly Charges, Total Charges', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x='Contract', y='MonthlyCharges', data=df, hue='Contract', palette='Set2', legend=False)
plt.title('Monthly Charges Distribution by Contract')
plt.xlabel('Contract Type')
plt.ylabel('Monthly Charges')
plt.show()
```

```python
payment_churn = pd.crosstab(df['PaymentMethod'], df['Churn'], normalize='index') * 100

payment_churn.plot(kind='bar', stacked=True, figsize=(10, 5), color=['#457b9d', '#e63946'])
plt.title('Churn Percentage by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=30)
plt.legend(title='Churn')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))
sns.lineplot(x='tenure', y='TotalCharges', hue='Churn', data=df, palette=['#2b2d42', '#d90429'])
plt.title('Total Charges Growth over Tenure by Churn Status')
plt.xlabel('Tenure (Months)')
plt.ylabel('Total Charges')
plt.show()
```
